```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

mat = pd.read_csv("student-mat.csv", sep=",")
por = pd.read_csv("student-por.csv", sep=",")

print("Maths :", mat.shape)
print("Portugais :", por.shape)
mat.head()
```

```python
# Variables d'intérêt principal
fig, axes = plt.subplots(2, 3, figsize=(14, 8))

# Notes G1, G2, G3
for i, col in enumerate(['G1', 'G2', 'G3']):
    axes[0, i].hist(mat[col], bins=20, color='steelblue', edgecolor='white')
    axes[0, i].set_title(f'Distribution {col} (maths)')
    axes[0, i].set_xlabel('Note /20')
    axes[0, i].axvline(mat[col].mean(), color='red', linestyle='--', label='moyenne')
    axes[0, i].legend()

# Alcool semaine, weekend, et absences
axes[1, 0].hist(mat['Dalc'], bins=5, color='coral', edgecolor='white')
axes[1, 0].set_title('Alcool semaine (Dalc)')
axes[1, 0].set_xlabel('1 (faible) → 5 (fort)')

axes[1, 1].hist(mat['Walc'], bins=5, color='salmon', edgecolor='white')
axes[1, 1].set_title('Alcool weekend (Walc)')
axes[1, 1].set_xlabel('1 (faible) → 5 (fort)')

axes[1, 2].hist(mat['absences'], bins=20, color='mediumpurple', edgecolor='white')
axes[1, 2].set_title('Absences')
axes[1, 2].set_xlabel('Nombre de jours')

plt.tight_layout()
plt.savefig('distributions.png', dpi=150)
plt.show()
```
